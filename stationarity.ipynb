# Stationarity in Time Series

A time series is said to be stationary when its statistical properties do not change over time. A stationary series “looks statistically the same” no matter where you observe it in time.

`In simple words, "A stationary time series doesn’t change its behavior, only the numbers move up and down in a similar way".`

In practical terms, a time series is stationary if it has following properties:

- Mean remains constant over time (For example, average temperature stays around 30°C every month)
- Variance (Spread) remains constant over time (The ups and downs don’t get bigger or smaller over time. For example, temperatures move between 28°C and 32°C every year)
- Autocorrelation structure (relationship with its past values) remains constant over time (How today’s value depends on yesterday’s value does not change. For example, if today is hot, tomorrow is usually hot — and this rule stays true)

**Why Stationarity matters:** Because many math and computer models only work well if the data is stationary. If the data keeps changing its behavior, the model gets confused. Most classical time-series modesl (ARMA, ARIMA, SARIMA) assumes stationarity.

# Types of Stationarity

## Weak Stationarity

- Constant mean, variance and auto correlation.
- This is very common.
- Example: Temperature in a classroom during school days.
  - Avg temperature remains the same around $25^oC$
  - Temperature varies between 23 to 27
  - If today is warm tomorrow is likely warm too
  - These patterns stay similar over months
  
## Strict Stationarity

- Exhibt the properties of weak stationarity
- The entire probability distribution of the data does not change when shifted in time. In simple words, not just the average and spread but everything about the data stays the same, no matter when you look at it.
- This type of stationarity is rare.
- Example: A perfectly controlled laboratory room
  - Everyday, temperature follows exactly the same probability pattern
  - Not just the avg or range, the entire behavior is identical
  - If you shift the day by 1 day, month or year, nothing change statistically

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

## No Stationarity (Random)

```python
# Random walk (non-stationary)
random_walk = np.cumsum(np.random.normal(0, 1, 50))

rw = pd.Series(random_walk)

print("Mean:", rw.mean())
print("Variance:", rw.var())
```

```python
rw.plot(title="Non-Stationary Time Series (Random Walk)")
plt.show()
```

## Weak Stationarity

```python
# Generate white noise (classic weakly stationary series)
data = np.random.normal(loc=0, scale=1, size=50)

wts = pd.Series(data)

print("Mean:", wts.mean())
print("Variance:", wts.var())
```

```python
wts.plot(title="Weakly Stationary Time Series (White Noise)")
plt.show()
```

## Strict Stationarity

This is theoritically only and it's very rare.

```python
# Same distribution at all times (theoretical strong stationarity)
strong_ts = np.random.normal(0, 1, 50)

sts = pd.Series(strong_ts)

print("Mean:", sts.mean())
print("Variance:", sts.var())
```

```python
sts.plot(title="Strict Stationary Time Series")
plt.show()
```
